## Stuff document chain

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("../test_pdfs/attention.pdf")
docs = loader.load()
```

```python
docs
```

```python
template = """
    Write a concise and short summary of the following research paper:{text}
"""
prompt = PromptTemplate(input_variables=["text"],template=template)
```

```python
from langchain.chains.summarize import load_summarize_chain
chain = load_summarize_chain(llm=llm,verbose= True, prompt= prompt, chain_type="stuff")
```

```python
res  = chain.run(docs)
```

```python
print(res)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 2000, chunk_overlap = 100)
final_docs = text_splitter.split_documents(docs)
```

```python
chunk_promp = """
    Please summarize this paper: {text}
"""
map_prompt = PromptTemplate(input_variables=["text"],template=chunk_promp)
```

```python
final_prompt_template = """
    Give the final summary of the entire research paper with these important points:\
        - description on the math behind it 
        - a motivational title
        -Start the precise summary with an introduction
        - Provide the summary in numbered points for the paper.
        
        {text}
        Also, note that i want u to response it in a bit of fun charismatic , witty way so that it would feel more fun
    
"""
final_prompt = PromptTemplate(input_variables=["text"],template=final_prompt_template)
```

```python
summary_chain = load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    # chain_type="refine"
    combine_prompt = final_prompt,
    verbose=True
)
```

```python
response = summary_chain.run(final_docs)
```

```python
print(response)
```
